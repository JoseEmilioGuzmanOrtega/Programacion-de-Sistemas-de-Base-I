- *¿Cuáles son las fases de la etapa de Análisis (Front-End) de un compilador?*  
**RESPUESTA:**
La etapa de Análisis (Front-End) de un compilador se compone de cuatro fases principales:

  **Análisis Léxico (o Lexicográfico):** Lee el código fuente carácter por carácter, elimina comentarios y espacios en blanco, y agrupa los caracteres en unidades con significado lógico llamadas tokens o componentes léxicos.


  **Análisis Sintáctico:** Toma los tokens de la fase anterior y verifica que su orden cumpla con las reglas gramaticales del lenguaje, construyendo una estructura jerárquica llamada árbol sintáctico.


  **Análisis Semántico:** Revisa el significado lógico del programa para asegurar que tenga sentido (por ejemplo, realiza el chequeo de compatibilidad de tipos y verifica que las variables hayan sido declaradas antes de usarse).


  **Generación de Código Intermedio:** Traduce el programa validado a una representación intermedia o abstracta (independiente de la máquina física), facilitando la posterior traducción y optimización en el Back-End.


  Adicionalmente, de forma transversal a todas las fases del front-end, operan dos elementos esenciales.


  **La Tabla de Símbolos:** Almacena la información de todos los identificadores declarados y sus atributos asociados.


  **El Manejo de Errores:** Detecta, reporta y se recupera de los fallos encontrados en cada nivel del análisis.

- *¿Qué diferencia a un Lexema de un Token?*  
**RESPUESTA:** La diferencia fundamental entre un Lexema y un Token radica en que uno representa el texto real escrito en el código y el otro representa su categoría lógica dentro del lenguaje:

  **Lexema:** Es la secuencia concreta de caracteres (la "palabra" física) que el analizador léxico lee en el programa fuente y que coincide con un patrón determinado.

      Ejemplos: x, comision, 3.14, while, +

  **Token:** Es el símbolo abstracto o la categoría lógica que el compilador le asigna a ese lexema para que el analizador sintáctico pueda trabajar con él. Internamente, suele representarse como un número entero o un identificador único.

      Ejemplos: ID (para identificar variables), NUM (para constantes numéricas), OP_SUMA o palabras clave como WHILE

- *¿Qué función cumple la Tabla de Símbolos durante la compilación?*  
**RESPUESTA:** La Tabla de Símbolos funciona como el "diccionario central" del compilador. Su única gran tarea es almacenar y organizar los nombres de los identificadores (variables, funciones, constantes) que escribes en tu código para que el compilador pueda:

  **Registrar:** Guardar el nombre de cada variable o función en cuanto aparece en el código.

  **Validar:** Comprobar que las variables hayan sido declaradas antes de usarse, que no se repitan nombres de forma ilegal y que sus tipos de datos sean compatibles.

  **Traducir:** Guardar la dirección de memoria de cada variable para que el compilador sepa exactamente de dónde leer o escribir los datos en el programa final.
  
  En resumen, es el mapa que le dice al compilador qué significa cada nombre en tu código, dónde se puede usar y en qué parte de la memoria se guarda.

## El "Monstruo Completo" (Enlace Estático)  
Por defecto, GCC usa ***enlace dinámico*** usando una estructura de datos tabular, la tabla **PLT** (*Procedure Linkage Table*) ya que el código usa `printf`, el programa necesita conectarse con la biblioteca estándar de C (`libc.so`). El enlazador no copia el código de `printf` dentro del archivo para no duplicar espacio en el disco; en su lugar, crea un "trampolín" o acceso directo en una sección llamada `<printf@plt>`.

Cada vez que se invoca a `printf`, realmente salta a este bloque plt, el cual averigua en qué parte de la memoria RAM del sistema operativo se encuentra la verdadera función `printf` y redirige el control allí en tiempo de ejecución.

Para ver cómo el enlazador inyecta literalmente miles y miles de líneas copiando físicamente el código de `printf` y todas sus dependencias dentro del ejecutable, se debe compilar de forma estática:

```bash
# Compilación estática

````python
import token
import tokenize
from io import BytesIO

# Código fuente de prueba en formato cadena
codigo_fuente = "suma = a + 10"

# Tokenización del código fuente
tokens = tokenize.tokenize(BytesIO(codigo_fuente.encode("utf-8")).readline)

print(
    f"{'LINEA/COL':<12} | {'TIPO DE TOKEN':<20} | {'VALOR (LEXEMA)':<15}"
)
print("-" * 55)
for tok in tokens:
    if tok.type in (
        tokenize.ENCODING,
        tokenize.ENDMARKER,
        tokenize.NL,
        tokenize.NEWLINE,
    ):
        continue
    nombre_token = token.tok_name[tok.type]
    posicion = f"{tok.start[0]}:{tok.start[1]}"
    print(f"{posicion:<12} | {nombre_token:<20} | {tok.string:<15}")
````

### [CÓDIGO 3.1] Simulación de una Tabla de Símbolos básica en Python
```python
class TablaDeSimbolos:

    def __init__(self):
        self.simbolos = {}

    def insertar(self, nombre, tipo, valor=None, ambito="global"):
        if nombre in self.simbolos:
            print(f"[ERROR LÉXICO/SINTÁCTICO] Identificador '{nombre}' ya declarado.")
        else:
            self.simbolos[nombre] = {
                "tipo": tipo,
                "valor": valor,
                "ambito": ambito,
            }
            print(f"[TABLA DE SÍMBOLOS] Insertado: {nombre} ({tipo})")

    def buscar(self, nombre):
        return self.simbolos.get(nombre, None)

    def mostrar(self):
        print("\n=== CONTENIDO DE LA TABLA DE SÍMBOLOS ===")
        print(f"{'NOMBRE':<12} | {'TIPO':<10} | {'ÁMBITO':<10} | {'VALOR':<10}")
        print("-" * 50)
        for nombre, datos in self.simbolos.items():
            print(
                f"{nombre:<12} | {datos['tipo']:<10} | {datos['ambito']:<10} | {str(datos['valor']):<10}"
            )

2. Implementa una función de prueba preliminar (o utiliza la librería `tokenize` de Python como simulador) que tome la cadena de tu lenguaje y la desglose en una lista de componentes léxicos.

````python
import tokenize
import io


def tokenizar_topored(ruta_archivo):
    """
    Prototipo de inspeccion lexica para archivos .red
    Utiliza la libreria tokenize de Python como aproximacion inicial.
    """
    tokens_encontrados = []
    with open(ruta_archivo, 'r', encoding='utf-8') as f:
        codigo = f.read()
    buffer = io.StringIO(codigo)
    try:
        for tok in tokenize.generate_tokens(buffer.readline):
            if tok.type in (tokenize.NEWLINE, tokenize.NL,
                            tokenize.INDENT, tokenize.DEDENT,
                            tokenize.ENDMARKER):
                continue
            tokens_encontrados.append({
                'tipo': tokenize.tok_name[tok.type],
                'valor': tok.string,
                'linea': tok.start[0],
                'columna': tok.start[1]
            })
    except tokenize.TokenError as e:
        print(f"Error al tokenizar: {e}")

    return tokens_encontrados


def mostrar_tokens(tokens):
    print(f"{'#':<4} {'TIPO':<12} {'VALOR':<20} {'LINEA':<6} {'COL'}")
    print('-' * 55)
    for i, t in enumerate(tokens, 1):
        print(f"{i:<4} {t['tipo']:<12} {t['valor']:<20} "
              f"{t['linea']:<6} {t['columna']}")


# Ejecucion del prototipo
tokens = tokenizar_topored('topologia_prueba.red')
mostrar_tokens(tokens)
````

3. Muestra una estructura en código que sirva como **Tabla de Símbolos inicial** donde se registren las variables declaradas en tu lenguaje.

````python
class Dispositivo:
    """
    Estructura que representa un simbolo (dispositivo) dentro
    de la tabla de simbolos de TopoRed.
    """

    def __init__(self, nombre, tipo, atributos=None):
        self.nombre = nombre.lower()      # No distingue mayus/minus
        self.tipo = tipo                  # router, switch, modem, wifi
        self.atributos = atributos or {}  # ip, ssid, password, etc.
        self.conexiones = []              # dispositivos conectados

    def __repr__(self):
        return f"<{self.tipo} {self.nombre} atributos={self.atributos}>"


class TablaSimbolos:
    """
    Registra los dispositivos declarados dentro de un bloque 'red'.
    """

    def __init__(self, nombre_red):
        self.nombre_red = nombre_red
        self.simbolos = {}   # clave: nombre en minusculas

    def declarar(self, nombre, tipo, atributos=None):
        clave = nombre.lower()
        if clave in self.simbolos:
            print(f"Error: '{nombre}' ya fue declarado.")
            return False
        self.simbolos[clave] = Dispositivo(nombre, tipo, atributos)
        return True

    def existe(self, nombre):
        return nombre.lower() in self.simbolos

    def obtener(self, nombre):
        return self.simbolos.get(nombre.lower())

    def eliminar(self, nombre):
        clave = nombre.lower()
        if clave in self.simbolos:
            del self.simbolos[clave]
            return True
        print(f"Error: '{nombre}' no existe.")
        return False
````

**RESPUESTA:**

Sí, hay un conflicto ya que las palabras reservadas (router, switch, tipo, donde, y, o, etc.) tienen exactamente la misma forma que un identificador válido (letra + letras/dígitos/guiones bajos), porque ninguna lleva algo que las distinga a simple vista. O sea, si alguien nombra un dispositivo tipo o y, el analizador no sabría a primera vista si es una palabra clave o un nombre inventado por el usuario.

También hay un mini-conflicto extra: Como los identificadores no distinguen mayúsculas de minúsculas, alguien podría intentar usar Router pensando que así se libra de la restricción.

**Cómo se resuelve:** El analizador primero reconoce cualquier palabra con ese patrón (letra + letras/dígitos/_) sin importar si es reservada o no. Ya que la tiene completa, la pasa a minúsculas y la busca en una lista/diccionario de palabras reservadas.

              Si está en la lista → se marca como esa palabra reservada.

                    Si no está → se marca como identificador normal.

Así, con un solo paso y una simple búsqueda en tabla, se evita que alguien use una palabra reservada como nombre de dispositivo, sin necesidad de reglas complicadas.

2. **Pregunta 2 (Estructura de la Tabla de Símbolos):** De los elementos de tu lenguaje original, ¿cuáles atributos (tipo, valor, ámbito, dirección de memoria) necesitará almacenar tu Tabla de Símbolos cuando se procese una declaración?

**RESPUESTA:**

En TopoRed, cuando se procesa una declaración (crear router R1 ip "10.0.0.1"), la Tabla de Símbolos necesita guardar esto por cada dispositivo:

- **Nombre:** El identificador, guardado en minúsculas (porque R1 y r1 son el mismo).

- **Tipo:** Qué clase de equipo es (router, switch, modem, wifi o dispositivo), para poder validar después cosas como "que solo se filtre por tipo == router".

- **Atributos y sus valores:** Cada dispositivo trae distintos atributos según su tipo (ip, ssid, password), así que se guardan como una lista de pares atributo-valor.

- **Ámbito:** A qué bloque red pertenece (por ejemplo, "Campus Norte"), ya que todo está declarado dentro de ese bloque con nombre.

- **Conexiones:** Qué otros dispositivos tiene enlazados y con qué ancho de banda, porque eso se define después con conectar y hay que poder consultarlo o modificarlo.

Lo que no aplica aquí es "dirección de memoria", porque TopoRed no genera código máquina ni maneja memoria física — es un lenguaje declarativo que describe una topología, no un programa que se ejecuta con variables en RAM. Eso sí tendría sentido en un compilador tradicional, pero no en este caso.
